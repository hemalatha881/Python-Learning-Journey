```python
# Create a string made of the middle three characters
# Write a program to create a new string made of the middle three characters of an input string.
# ex.JaSonAy-->Son
# JhonDipPeta -->Dip

st= input("Enter a string")
m = len(st)
print("length of string",m)
if m %2 ==0:
    print("not possible")
else:
    c = len(st)//2
    x = st[c-1]
    y = st[c]
    z = st[c+1]
    op = ''.join([x,y,z])
    print("middle three characters " ,op)


```

```python
#remove characters from string
s = input("Enter a string ")
c = input("which character should be remove : ")
l = []
for i in s:
    if i != c:
        l.append(i)
    else:
        continue

sr = ''.join(l)
print(sr)
```
