```python
#to create seperate prime numbers list ,even number list and odd number list of given range

start = int(input("Enter first value"))
end = int(input("Enter last value"))
lis = []
ev = []
od = []
count = 0
for i in range(start,end+1):

    if i > 1:
        
        for j in range (2,i):
            
            if (i % j ==0):
                break
        else:
            lis.append(i)
    
    
    if(i%2==0):
        ev.append(i)
    else:
        od.append(i)
        
        
print("prime numbers: ",lis)
print("even numbers: ",ev)
print("odd numbers: ",od)
```

```python
st = "My NaMe is jj,I'm 30 years old!"
#seperate list of upper case,lower case,number,special characters

uc = []
lc = []
n = []
sc = []
for i in st:
    if i == " ":
        continue
    if i.isupper():
        uc.append(i)
    elif i.islower():
        lc.append(i)
    elif i.isdigit():
        n.append(i)
    else:
        sc.append(i)
print(f"{uc}\n{lc}\n{n}\n{sc}")
```
